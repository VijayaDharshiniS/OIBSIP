```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

apps = pd.read_csv("googleplaystore.csv")
reviews = pd.read_csv("googleplaystore_user_reviews.csv")

print("Apps dataset shape:", apps.shape)
print("Reviews dataset shape:", reviews.shape)

print("\nApps columns:")
print(apps.columns.tolist())

print("\nReviews columns:")
print(reviews.columns.tolist())

print("\nApps dataset preview:")
display(apps.head())

print("\nReviews dataset preview:")
display(reviews.head())
```

## 2. Data Cleaning and Preprocessing

The Google Play Store datasets contain real-world data that may include missing values, duplicate records, inconsistent data types, and formatted numeric values. The app dataset is inspected and cleaned by converting columns such as Installs, Price, and Size into appropriate numeric formats. Missing values and duplicates are handled carefully. The user reviews dataset is also checked to prepare it for sentiment analysis.

```python

print("Apps dataset information:")
apps.info()

print("\nMissing values in apps dataset:")
display(apps.isnull().sum().to_frame("Missing Values"))

print("\nDuplicate rows in apps dataset:", apps.duplicated().sum())

print("\nMissing values in reviews dataset:")
display(reviews.isnull().sum().to_frame("Missing Values"))

print("\nDuplicate rows in reviews dataset:", reviews.duplicated().sum())

print("\nApp dataset data types:")
display(apps.dtypes.to_frame("Data Type"))
```

```python

apps = pd.read_csv("googleplaystore.csv")

apps = apps.drop_duplicates().copy()

apps["Rating"] = pd.to_numeric(apps["Rating"], errors="coerce")
apps["Reviews"] = pd.to_numeric(apps["Reviews"], errors="coerce")

apps["Installs"] = pd.to_numeric(
    apps["Installs"].astype(str)
    .str.replace(",", "", regex=False)
    .str.replace("+", "", regex=False),
    errors="coerce"
)

apps["Price"] = pd.to_numeric(
    apps["Price"].astype(str)
    .str.replace("$", "", regex=False),
    errors="coerce"
)

def convert_size(value):
    value = str(value).strip().upper()

    if value in ["NAN", "VARIES WITH DEVICE"]:
        return np.nan

    try:
        if value.endswith("M"):
            return float(value[:-1])
        if value.endswith("K"):
            return float(value[:-1]) / 1024
        return np.nan
    except ValueError:
        return np.nan

apps["Size_MB"] = apps["Size"].apply(convert_size)

apps["Last Updated"] = pd.to_datetime(
    apps["Last Updated"],
    errors="coerce"
)

apps["Type"] = apps["Type"].fillna(
    apps["Price"].apply(
        lambda x: "Paid" if pd.notna(x) and x > 0 else "Free"
    )
)

print("Cleaned apps dataset shape:", apps.shape)
print("\nColumn data types:")
display(apps[["Rating", "Reviews", "Size_MB", "Installs", "Price"]].dtypes.to_frame("Data Type"))

print("\nCleaned data preview:")
display(apps[["App", "Rating", "Size", "Size_MB", "Installs", "Price"]].head())

print("\nRemaining duplicate rows:", apps.duplicated().sum())
```

```python

reviews = pd.read_csv("googleplaystore_user_reviews.csv")

reviews = reviews.drop_duplicates().copy()

reviews = reviews.dropna(subset=["Translated_Review"]).copy()

reviews["Translated_Review"] = reviews["Translated_Review"].astype(str)

print("Cleaned reviews dataset shape:", reviews.shape)

print("\nMissing values after cleaning:")
display(reviews.isnull().sum().to_frame("Missing Values"))

print("\nSample cleaned reviews:")
display(reviews[["App", "Translated_Review"]].head())
```

## 3. Category Analysis

Category analysis helps us understand how apps are distributed across the Google Play Store. A bar chart is used to compare the number of apps in each category. Categories with more listed apps may indicate stronger competition or a more crowded market, although app count alone does not prove market saturation or user demand.

```python

category_counts = apps["Category"].value_counts()

plt.figure(figsize=(14, 7))
sns.barplot(
    x=category_counts.values,
    y=category_counts.index
)

plt.title("Number of Apps Across Google Play Store Categories")
plt.xlabel("Number of Apps")
plt.ylabel("Category")
plt.tight_layout()
plt.show()

print("Top 10 categories by app count:")
display(category_counts.head(10).to_frame("Number of Apps"))
```

## 4. Ratings Analysis

Ratings provide an indication of user feedback about apps. This analysis examines the distribution of app ratings and compares average ratings across categories. Missing ratings are excluded from the rating calculations rather than being replaced with arbitrary values. Category averages should be interpreted alongside the number of rated apps, because categories with few ratings may have less reliable averages.

```python

rating_data = apps.dropna(subset=["Rating"])

plt.figure(figsize=(9, 5))
sns.histplot(rating_data["Rating"], bins=20, kde=True)

plt.title("Distribution of Google Play Store App Ratings")
plt.xlabel("App Rating")
plt.ylabel("Number of Apps")
plt.tight_layout()
plt.show()

print("Average app rating:", round(rating_data["Rating"].mean(), 2))
print("Median app rating:", round(rating_data["Rating"].median(), 2))
print("Number of apps with ratings:", len(rating_data))
```

```python

category_ratings = (
    apps.dropna(subset=["Rating"])
    .groupby("Category")
    .agg(
        Average_Rating=("Rating", "mean"),
        Rated_Apps=("Rating", "count")
    )
    .sort_values("Average_Rating", ascending=False)
)

plt.figure(figsize=(14, 7))
sns.barplot(
    data=category_ratings.reset_index(),
    x="Average_Rating",
    y="Category"
)

plt.title("Average App Rating by Category")
plt.xlabel("Average Rating")
plt.ylabel("Category")
plt.xlim(0, 5)
plt.tight_layout()
plt.show()

print("Top 10 categories by average rating:")
display(category_ratings.head(10).round(2))
```

## 5. App Size and Installs Analysis

This analysis explores the relationship between app size and the number of installs. A scatter plot is used to visualise the two variables, while Pearson correlation measures their linear relationship. Since the install counts vary widely, a logarithmic scale is used to make the plot easier to interpret. Correlation does not imply causation, and missing app sizes are excluded from this analysis.

```python

size_data = apps.dropna(subset=["Size_MB", "Installs"]).copy()

size_data = size_data[
    (size_data["Size_MB"] > 0) &
    (size_data["Installs"] > 0)
]

plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=size_data,
    x="Size_MB",
    y="Installs",
    alpha=0.4
)

plt.yscale("log")
plt.title("App Size vs Number of Installs")
plt.xlabel("App Size (MB)")
plt.ylabel("Number of Installs (log scale)")
plt.tight_layout()
plt.show()

correlation = size_data["Size_MB"].corr(size_data["Installs"])

print("Pearson correlation between app size and installs:", round(correlation, 4))
print("Apps included in this analysis:", len(size_data))
```

## 6. Free and Paid App Analysis

This analysis compares free and paid apps in the Google Play Store. It examines the distribution of app types, the prices of paid apps, and potential revenue across categories. The revenue calculation uses app price multiplied by the reported install count as a rough estimate only. Actual revenue cannot be determined from this dataset because installs do not necessarily represent purchases, and install counts may be rounded.

```python

apps["App_Type"] = np.where(apps["Price"] > 0, "Paid", "Free")

type_counts = apps["App_Type"].value_counts()

plt.figure(figsize=(7, 5))
sns.barplot(x=type_counts.index, y=type_counts.values)

plt.title("Distribution of Free and Paid Apps")
plt.xlabel("App Type")
plt.ylabel("Number of Apps")
plt.tight_layout()
plt.show()

display(type_counts.to_frame("Number of Apps"))

print("Free apps:", int((apps["App_Type"] == "Free").sum()))
print("Paid apps:", int((apps["App_Type"] == "Paid").sum()))
```

```python

paid_apps = apps[apps["Price"] > 0].copy()

plt.figure(figsize=(9, 5))
sns.histplot(data=paid_apps, x="Price", bins=30)

plt.title("Price Distribution of Paid Apps")
plt.xlabel("Price")
plt.ylabel("Number of Apps")
plt.tight_layout()
plt.show()

print("Number of paid apps:", len(paid_apps))
print("Average paid app price:", round(paid_apps["Price"].mean(), 2))
print("Median paid app price:", round(paid_apps["Price"].median(), 2))
```

```python

revenue_data = apps[
    (apps["Price"] > 0) &
    (apps["Installs"] > 0)
].copy()

revenue_data["Estimated_Revenue"] = (
    revenue_data["Price"] * revenue_data["Installs"]
)

category_revenue = (
    revenue_data.groupby("Category")["Estimated_Revenue"]
    .sum()
    .sort_values(ascending=False)
)

top_revenue = category_revenue.head(10).sort_values()

plt.figure(figsize=(11, 6))
top_revenue.plot(kind="barh")

plt.title("Top 10 Categories by Estimated Revenue")
plt.xlabel("Estimated Revenue (Price × Installs)")
plt.ylabel("Category")
plt.tight_layout()
plt.show()

display(category_revenue.head(10).to_frame("Estimated Revenue").round(2))
```

## 7. Sentiment Analysis of User Reviews

Sentiment analysis is used to understand users' opinions about apps by analysing their written reviews. TextBlob calculates a polarity score between -1 and +1. Reviews with positive polarity are classified as Positive, negative polarity as Negative, and zero polarity as Neutral. This provides an overview of user sentiment, although automated sentiment analysis may not fully understand sarcasm or context.

```python

%pip install textblob
```

```python

from textblob import TextBlob

reviews["Polarity"] = reviews["Translated_Review"].apply(
    lambda text: TextBlob(str(text)).sentiment.polarity
)

reviews["Sentiment_Label"] = np.where(
    reviews["Polarity"] > 0,
    "Positive",
    np.where(reviews["Polarity"] < 0, "Negative", "Neutral")
)

sentiment_counts = reviews["Sentiment_Label"].value_counts()

plt.figure(figsize=(8, 5))
sns.barplot(
    x=sentiment_counts.index,
    y=sentiment_counts.values
)

plt.title("Sentiment Distribution of App Reviews")
plt.xlabel("Sentiment")
plt.ylabel("Number of Reviews")
plt.tight_layout()
plt.show()

display(sentiment_counts.to_frame("Number of Reviews"))

print("Average review polarity:", round(reviews["Polarity"].mean(), 3))
```

## 8. Sentiment Analysis by App Category

This analysis compares positive, negative, and neutral user reviews across Google Play Store app categories. The review dataset is combined with the app dataset using the app name. Sentiment percentages are used to compare categories more fairly than raw review counts, because some categories contain more reviews than others.

```python

app_categories = apps[["App", "Category"]].drop_duplicates(subset=["App"])

reviews_category = reviews.merge(
    app_categories,
    on="App",
    how="left"
)

print("Reviews with category information:", len(reviews_category))
print("Reviews without category:", reviews_category["Category"].isna().sum())

display(reviews_category[["App", "Category", "Sentiment_Label"]].head())
```

```python

sentiment_by_category = pd.crosstab(
    reviews_category["Category"],
    reviews_category["Sentiment_Label"],
    normalize="index"
) * 100

sentiment_by_category = sentiment_by_category.dropna()

sentiment_by_category.plot(
    kind="bar",
    stacked=True,
    figsize=(15, 7)
)

plt.title("Sentiment Distribution by App Category")
plt.xlabel("App Category")
plt.ylabel("Percentage of Reviews")
plt.xticks(rotation=90)
plt.legend(title="Sentiment")
plt.tight_layout()
plt.show()

display(sentiment_by_category.round(2))
```

```python

if "Positive" in sentiment_by_category.columns:
    print("Top 5 categories by positive sentiment percentage:")
    display(
        sentiment_by_category["Positive"]
        .sort_values(ascending=False)
        .head(5)
        .round(2)
        .to_frame("Positive Reviews (%)")
    )

if "Negative" in sentiment_by_category.columns:
    print("Top 5 categories by negative sentiment percentage:")
    display(
        sentiment_by_category["Negative"]
        .sort_values(ascending=False)
        .head(5)
        .round(2)
        .to_frame("Negative Reviews (%)")
    )
```

## 9. Key Findings and Recommendations

The analysis of Google Play Store apps examines category distribution, ratings, app size, installs, pricing, estimated revenue potential, and user review sentiment.

### Key Findings

1. **Category competition:** The category distribution identifies which categories contain the largest number of apps. A high app count may indicate greater competition, but it does not directly measure market demand.

2. **Ratings and app size:** Rating analysis compares user ratings across categories, while the size-versus-installs analysis explores whether app size has a linear relationship with reported installs.

3. **Pricing and user sentiment:** The free-versus-paid comparison shows the distribution of pricing models. Review sentiment analysis provides an additional view of user opinions across app categories.

### Recommendations for App Developers

1. Study categories with relatively fewer apps and promising ratings before choosing a niche. Validate actual demand before development.

2. Optimise app size and performance for the target audience. Use install and rating data together rather than relying on a single metric.

3. Analyse negative reviews to identify common usability issues and prioritise improvements. Use positive reviews to understand which features users appreciate.

### Limitations

The dataset represents a snapshot of the Google Play Store rather than current live market conditions. Missing values, duplicate records, rounded install counts, and automated sentiment classification may affect the results. Estimated revenue calculated from price multiplied by installs is not actual revenue.
